```python
import math
import ipywidgets as widgets
from IPython.display import display, clear_output

# Labels: English, Kannada
LABELS = {
    "profit": ("Crop Profit", "ಬೆಳೆಯ ಲಾಭ"),
    "seed": ("Seed Requirements", "ಬೀಜದ ಅಗತ್ಯ"),
    "nursery": ("Nursery Pricing", "ನರ್ಸರಿ ಬೆಲೆ"),
    "market": ("Market Comparison", "ಮಾರುಕಟ್ಟೆ ಹೋಲಿಕೆ"),

    "harvest": ("Saleable harvest (kg)", "ಮಾರಾಟಯೋಗ್ಯ ಇಳುವರಿ (ಕೆಜಿ)"),
    "price": ("Selling price per kg", "ಪ್ರತಿ ಕೆಜಿ ಮಾರಾಟ ಬೆಲೆ"),
    "seeds": ("Seed cost", "ಬೀಜದ ವೆಚ್ಚ"),
    "fertiliser": ("Fertiliser cost", "ಗೊಬ್ಬರದ ವೆಚ್ಚ"),
    "labour": ("Labour cost", "ಕಾರ್ಮಿಕ ವೆಚ್ಚ"),
    "water": ("Irrigation cost", "ನೀರಾವರಿ ವೆಚ್ಚ"),
    "transport": ("Transport cost", "ಸಾಗಣೆ ವೆಚ್ಚ"),
    "other": ("Other costs", "ಇತರೆ ವೆಚ್ಚ"),

    "area": ("Plantable area (m²)", "ನೆಡುವ ವಿಸ್ತೀರ್ಣ (ಮೀ²)"),
    "row": ("Row spacing (cm)", "ಸಾಲುಗಳ ಅಂತರ (ಸೆಂಮೀ)"),
    "plant": ("Plant spacing (cm)", "ಸಸ್ಯಗಳ ಅಂತರ (ಸೆಂಮೀ)"),
    "germination": ("Germination (%)", "ಮೊಳಕೆಯ ಪ್ರಮಾಣ (%)"),

    "plants": ("Plants started", "ಆರಂಭಿಸಿದ ಸಸ್ಯಗಳ ಸಂಖ್ಯೆ"),
    "loss": ("Plant losses (%)", "ಸಸ್ಯ ನಷ್ಟ (%)"),
    "materials": ("Materials cost", "ಸಾಮಗ್ರಿಗಳ ವೆಚ್ಚ"),
    "markup": ("Markup on cost (%)", "ವೆಚ್ಚದ ಮೇಲೆ ಹೆಚ್ಚುವರಿ (%)"),

    "price_a": ("Market A: price per kg", "ಮಾರುಕಟ್ಟೆ A: ಪ್ರತಿ ಕೆಜಿ ಬೆಲೆ"),
    "fee_a": ("Market A: commission (%)", "ಮಾರುಕಟ್ಟೆ A: ಕಮಿಷನ್ (%)"),
    "transport_a": ("Market A: transport cost", "ಮಾರುಕಟ್ಟೆ A: ಸಾಗಣೆ ವೆಚ್ಚ"),
    "price_b": ("Market B: price per kg", "ಮಾರುಕಟ್ಟೆ B: ಪ್ರತಿ ಕೆಜಿ ಬೆಲೆ"),
    "fee_b": ("Market B: commission (%)", "ಮಾರುಕಟ್ಟೆ B: ಕಮಿಷನ್ (%)"),
    "transport_b": ("Market B: transport cost", "ಮಾರುಕಟ್ಟೆ B: ಸಾಗಣೆ ವೆಚ್ಚ"),

    "revenue": ("Revenue", "ಆದಾಯ"),
    "cost": ("Total expenses", "ಒಟ್ಟು ವೆಚ್ಚ"),
    "net": ("Profit / loss", "ಲಾಭ / ನಷ್ಟ"),
    "break_price": ("Break-even price per kg", "ವೆಚ್ಚ ಸರಿದೂಗಿಸಲು ಪ್ರತಿ ಕೆಜಿ ಬೆಲೆ"),
    "positions": ("Estimated planting positions", "ಅಂದಾಜು ನೆಡುವ ಜಾಗಗಳ ಸಂಖ್ಯೆ"),
    "seed_count": ("Estimated seeds to buy", "ಖರೀದಿಸಬೇಕಾದ ಅಂದಾಜು ಬೀಜಗಳು"),
    "saleable": ("Estimated saleable plants", "ಅಂದಾಜು ಮಾರಾಟಯೋಗ್ಯ ಸಸ್ಯಗಳು"),
    "unit_cost": ("Cost per saleable plant", "ಪ್ರತಿ ಮಾರಾಟಯೋಗ್ಯ ಸಸ್ಯದ ವೆಚ್ಚ"),
    "sell_price": ("Selling price with markup", "ಹೆಚ್ಚುವರಿ ಸೇರಿಸಿದ ಮಾರಾಟ ಬೆಲೆ"),
    "net_a": ("Market A: net proceeds", "ಮಾರುಕಟ್ಟೆ A: ನಿವ್ವಳ ಆದಾಯ"),
    "net_b": ("Market B: net proceeds", "ಮಾರುಕಟ್ಟೆ B: ನಿವ್ವಳ ಆದಾಯ"),
    "difference": ("Difference", "ವ್ಯತ್ಯಾಸ"),
    "better": ("Better market", "ಉತ್ತಮ ಮಾರುಕಟ್ಟೆ"),
}

FIELDS = {
    "profit": [
        "harvest", "price", "seeds", "fertiliser",
        "labour", "water", "transport", "other"
    ],
    "seed": ["area", "row", "plant", "germination"],
    "nursery": ["plants", "loss", "materials", "labour", "other", "markup"],
    "market": [
        "harvest", "price_a", "fee_a", "transport_a",
        "price_b", "fee_b", "transport_b"
    ],
}

NOTES = {
    "profit": (
        "Use one crop cycle and one currency. Include all expenses.",
        "ಒಂದೇ ಬೆಳೆ ಅವಧಿ ಮತ್ತು ಕರೆನ್ಸಿ ಬಳಸಿ. ಎಲ್ಲಾ ವೆಚ್ಚಗಳನ್ನು ಸೇರಿಸಿ."
    ),
    "seed": (
        "Assumes one seedling per position. Exclude paths from area.",
        "ಪ್ರತಿ ಜಾಗಕ್ಕೆ ಒಂದು ಸಸಿ ಎಂದು ಊಹಿಸಲಾಗಿದೆ. ದಾರಿಗಳ ವಿಸ್ತೀರ್ಣ ಸೇರಿಸಬೇಡಿ."
    ),
    "nursery": (
        "Markup is added to cost; it is not profit margin.",
        "ವೆಚ್ಚದ ಮೇಲೆ ಹೆಚ್ಚುವರಿ ಸೇರಿಸಲಾಗುತ್ತದೆ; ಇದು ಲಾಭದ ಮಾರ್ಜಿನ್ ಅಲ್ಲ."
    ),
    "market": (
        "Net proceeds exclude production costs.",
        "ನಿವ್ವಳ ಆದಾಯದಲ್ಲಿ ಉತ್ಪಾದನಾ ವೆಚ್ಚ ಕಡಿತಗೊಳಿಸಿಲ್ಲ."
    ),
}

language = widgets.Dropdown(
    options=[("English", 0), ("ಕನ್ನಡ", 1)],
    description="Language:"
)
calculator = widgets.Dropdown()
form = widgets.VBox()
output = widgets.Output()
button = widgets.Button(
    description="Calculate",
    button_style="success",
    layout=widgets.Layout(width="180px", height="42px")
)
inputs = {}


def text(key):
    return LABELS[key][language.value]


def calculate(kind, v):
    # Validate numbers before calculating.
    if any(not math.isfinite(n) or n < 0 for n in v.values()):
        raise ValueError(
            "Enter finite, non-negative numbers."
            if language.value == 0
            else "ಶೂನ್ಯ ಅಥವಾ ಅದಕ್ಕಿಂತ ಹೆಚ್ಚಿನ ಸರಿಯಾದ ಅಂಕಿಗಳನ್ನು ನಮೂದಿಸಿ."
        )

    for key in ("harvest", "area", "row", "plant", "plants", "germination"):
        if key in v and v[key] <= 0:
            raise ValueError(text(key) + ": > 0")

    for key in ("germination", "loss", "fee_a", "fee_b"):
        if key in v and v[key] > 100:
            raise ValueError(text(key) + ": 0–100")

    if "plants" in v and not v["plants"].is_integer():
        raise ValueError(
            "Plants started must be a whole number."
            if language.value == 0
            else "ಸಸ್ಯಗಳ ಸಂಖ್ಯೆ ಪೂರ್ಣಾಂಕವಾಗಿರಬೇಕು."
        )

    if kind == "profit":
        cost = sum(v[k] for k in (
            "seeds", "fertiliser", "labour", "water", "transport", "other"
        ))
        revenue = v["harvest"] * v["price"]
        result = {
            "revenue": revenue,
            "cost": cost,
            "net": revenue - cost,
            "break_price": cost / v["harvest"],
        }

    elif kind == "seed":
        positions = v["area"] / (
            (v["row"] / 100) * (v["plant"] / 100)
        )
        result = {
            "positions": math.floor(positions),
            "seed_count": math.ceil(
                positions * 100 / v["germination"]
            ),
        }

    elif kind == "nursery":
        saleable = math.floor(v["plants"] * (100 - v["loss"]) / 100)

        if saleable < 1:
            raise ValueError(
                "No saleable plants remain. Adjust plant count or losses."
                if language.value == 0
                else "ಮಾರಾಟಕ್ಕೆ ಸಸ್ಯಗಳು ಉಳಿಯುವುದಿಲ್ಲ. ಸಂಖ್ಯೆ ಅಥವಾ ನಷ್ಟ ಬದಲಾಯಿಸಿ."
            )

        cost = v["materials"] + v["labour"] + v["other"]
        unit_cost = cost / saleable
        sell_price = unit_cost * (1 + v["markup"] / 100)

        result = {
            "saleable": saleable,
            "cost": cost,
            "unit_cost": unit_cost,
            "sell_price": sell_price,
            "net": sell_price * saleable - cost,
        }

    else:
        net_a = (
            v["harvest"] * v["price_a"] * (100 - v["fee_a"]) / 100
            - v["transport_a"]
        )
        net_b = (
            v["harvest"] * v["price_b"] * (100 - v["fee_b"]) / 100
            - v["transport_b"]
        )

        equal = "Equal" if language.value == 0 else "ಸಮಾನ"
        result = {
            "net_a": net_a,
            "net_b": net_b,
            "difference": abs(net_a - net_b),
            "better": "A" if net_a > net_b else "B" if net_b > net_a else equal,
        }

    if any(
        isinstance(n, (int, float)) and not math.isfinite(n)
        for n in result.values()
    ):
        raise ValueError(
            "Values are too large."
            if language.value == 0 else "ಅಂಕಿಗಳು ತುಂಬಾ ದೊಡ್ಡದಾಗಿವೆ."
        )

    return result


def build_form(change=None):
    previous = {key: field.value for key, field in inputs.items()}
    inputs.clear()
    children = []

    note = widgets.Label(value=NOTES[calculator.value][language.value])
    children.append(note)

    for key in FIELDS[calculator.value]:
        field = widgets.FloatText(
            value=previous.get(key, 0),
            layout=widgets.Layout(width="220px")
        )
        field.observe(clear_result, names="value")
        inputs[key] = field
        children.append(widgets.VBox([
            widgets.Label(value=text(key)),
            field
        ]))

    form.children = children
    button.description = "Calculate" if language.value == 0 else "ಲೆಕ್ಕ ಹಾಕಿ"
    output.clear_output()


def clear_result(change=None):
    output.clear_output()


def update_language(change=None):
    selected = calculator.value or "profit"
    calculator.options = [(text(key), key) for key in FIELDS]
    calculator.value = selected
    build_form()


def run_calculation(button_clicked):
    with output:
        clear_output()
        try:
            values = {key: field.value for key, field in inputs.items()}
            results = calculate(calculator.value, values)

            for key, value in results.items():
                formatted = (
                    f"{value:,}" if isinstance(value, int)
                    else f"{value:,.2f}" if isinstance(value, float)
                    else value
                )
                print(f"{text(key)}: {formatted}")

            print(
                "\nPlanning estimate only. Actual figures may differ."
                if language.value == 0
                else "\nಇದು ಅಂದಾಜು ಮಾತ್ರ. ನಿಜವಾದ ಅಂಕಿಗಳು ಬದಲಾಗಬಹುದು."
            )

        except (ValueError, OverflowError) as error:
            print(error)


calculator.observe(build_form, names="value")
language.observe(update_language, names="value")
button.on_click(run_calculation)

update_language()
display(widgets.VBox([language, calculator, form, button, output]))
```
